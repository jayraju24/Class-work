```python
#b. Apply Min-Max Scaling to: [10, 20, 30]

```

```python
data_min_max = [10, 20, 30]

X_min = min(data_min_max)
X_max = max(data_min_max)

scaled_data_min_max = []
for x in data_min_max:
    x_scaled = (x - X_min) / (X_max - X_min)
    scaled_data_min_max.append(x_scaled)

print(f"Original data: {data_min_max}")
print(f"Min-Max Scaled data: {scaled_data_min_max}")
```

```python
#c. Apply Z-score Standardization to: [5, 15, 25]
```

```python

data_z_score = [5, 15, 25]

mean_val = np.mean(data_z_score)
std_val = np.std(data_z_score)

scaled_data_z_score = []
for x in data_z_score:
    x_scaled = (x - mean_val) / std_val
    scaled_data_z_score.append(x_scaled)

print(f"Original data: {data_z_score}")
print(f"Mean: {mean_val:.2f}")
print(f"Standard Deviation: {std_val:.2f}")
print(f"Z-score Standardized data: {scaled_data_z_score}")
```

```python
#c. Convert: ["Small", "Medium", "Large"] using Ordinal Encoding.
```

```python
from sklearn.preprocessing import OrdinalEncoder
import numpy as np

categories_to_encode = ["Small", "Medium", "Large"]

# Define the order of the categories
# This is crucial for Ordinal Encoding to maintain the meaning
ordered_categories = np.array(["Small", "Medium", "Large"]).reshape(-1, 1)

# Initialize the OrdinalEncoder with the specified order
encoder = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])

# Fit and transform the data
# The input to fit_transform needs to be 2D, hence .reshape(-1, 1)
encoded_data = encoder.fit_transform(np.array(categories_to_encode).reshape(-1, 1))

print(f"Original categories: {categories_to_encode}")
print(f"Ordinal Encoded data: {encoded_data.flatten()}")
```

```python
# b. Write code to scale a numeric list using StandardScaler.

from sklearn.preprocessing import StandardScaler
import numpy as np

# Sample numeric list
numeric_list = [10, 50, 20, 80, 30]

# Reshape the data as StandardScaler expects a 2D array
numeric_list_reshaped = np.array(numeric_list).reshape(-1, 1)

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the data
scaled_list = scaler.fit_transform(numeric_list_reshaped)

print(f"Original numeric list: {numeric_list}")
print(f"Scaled list (StandardScaler): {scaled_list.flatten()}")
```

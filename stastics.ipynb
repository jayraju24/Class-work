```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(101)
#simulated dataset

data={
    "Math_score":np.random.normal(10,70,100),
    "science_score":np.random.normal(12,72,100),
    "Hours_study":np.random.normal(2,5,100)
}
df=pd.DataFrame(data)
print(df.head(),"\n")
```

```python
mean_math=df["Math_score"].mean()
mean_science=df["science_score"].mean()
mode_math=df["Math_score"].mode()
mode_science=df["science_score"].mode()
print(mean_math)
print(mean_science)
print(mode_math)
print(mode_science)


p25=np.percentile(df["Math_score"],25)
p50=np.percentile(df["Math_score"],50)
p75=np.percentile(df["Math_score"],75)
iqr=p75-p25

print("==== percentiles & Quartiles ====")
print("25th percentile:",p25)
print("50th percentile:",p50)
print("75th percentile:",p75)
print("iqr :",iqr,"\n")
```

```python
#data distribution
plt.figure(figsize=(10,6))
plt.hist(df["Math_score"],bins=20,color="red",edgecolor="black")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Distribution of Math Scores")
plt.show()
```
